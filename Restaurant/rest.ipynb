### Outlier Imputation Strategy
- Use IQR-based winsorization to cap extreme values.
- Optionally apply z-score clipping with median replacement.
- For skewed columns, consider log-transform before imputation.
- Group-wise capping (e.g., by City) preserves local distributions.
- Keep an audit of replacements to quantify impact.
